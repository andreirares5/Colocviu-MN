```python
import numpy as np
import pandas as pd

def generate_xg(distance, angle, body):
    # logistică pe distanță
    xg_dist = 1 / (1 + np.exp(0.25 * (distance - 11)))

    # penalizare pe unghi
    xg_angle = np.cos(np.radians(angle))

    # penalizare pe parte corp
    body_factor = 1.0 if body == 0 else 0.7

    xg = xg_dist * xg_angle * body_factor

    # clamp în [0,1]
    return max(0.0, min(1.0, xg))


# generăm dataset
N = 5000
distances = np.random.uniform(3, 30, N)
angles = np.random.uniform(0, 60, N)
body_parts = np.random.choice([0, 1], N)

xg_values = [
    generate_xg(d, a, b)
    for d, a, b in zip(distances, angles, body_parts)
]

df = pd.DataFrame({
    "distance": distances,
    "angle": angles,
    "body_part": body_parts,
    "xg": xg_values
})

df.to_csv("../data/shots.csv", index=False)
df.head()
```
