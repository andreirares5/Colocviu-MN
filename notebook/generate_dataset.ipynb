```python
import numpy as np
import pandas as pd

def generate_synthetic_xg_dataset(n_samples=2000, noise_std=0.02, seed=42):
    np.random.seed(seed)

    # 1. Generăm variabilele de intrare
    distance = np.random.uniform(5, 30, n_samples)          # metri
    angle = np.random.uniform(0, 60, n_samples)             # grade
    body_part = np.random.choice([0, 1], n_samples)         # 0 = picior, 1 = cap

    # 2. Modelul "adevărat" al probabilității
    true_prob = (
        0.6 * np.exp(-0.12 * distance) *
        np.cos(np.deg2rad(angle)) +
        0.05 * body_part +
        np.random.normal(0, noise_std, n_samples)
    )

    # 3. Clipping pentru a rămâne în [0,1]
    true_prob = np.clip(true_prob, 0, 1)

    # 4. Generăm eticheta (gol / nu)
    goals = (np.random.rand(n_samples) < true_prob).astype(int)

    # 5. Construim DataFrame
    df = pd.DataFrame({
        "distance": distance,
        "angle": angle,
        "body_part": body_part,
        "goal": goals,
        "true_prob": true_prob
    })

    return df


# Generăm dataset-ul și îl salvăm
df = generate_synthetic_xg_dataset()
df.to_csv("../data/shots.csv", index=False)

df.head()
```
